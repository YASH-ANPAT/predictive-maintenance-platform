Model Selection

Four machine learning algorithms were evaluated for predictive maintenance:

Logistic Regression
Decision Tree
Random Forest
XGBoost

Evaluation was based on:

Accuracy
Precision
Recall
F1 Score

Since predictive maintenance prioritizes detecting equipment failures, Recall and F1 Score were considered more important than Accuracy alone.

XGBoost achieved:

Highest Accuracy
Highest Recall
Highest F1 Score

Therefore, XGBoost was selected as the final production model.
